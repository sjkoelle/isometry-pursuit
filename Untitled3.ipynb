```python
import json
import numpy as np
import pandas as pd

path = "results/ethanol_results.json"

with open(path) as f:
    data = json.load(f)

rows = []
for r in data["replicates"]:
    greedy = r["greedy_gl_norm"]
    restricted = r["greedy_gl_bp_norm"]
    ip_support_loss = r["group_lasso_norm"]

    rows.append({
        "replicate": r["replicate"],
        "D": data["metadata"]["data_tensor_shape"][0],
        "P": data["metadata"]["data_tensor_shape"][1],
        "ip_support_size": r["support_cardinality"],
        "ip_support": r["basis_pursuit_indices"],
        "greedy_full_indices": r["greedy_gl_indices"],
        "greedy_restricted_indices": r["greedy_gl_bp_indices"],
        "greedy_full_loss": greedy,
        "greedy_restricted_loss": restricted,
        "ip_support_loss": ip_support_loss,
        "full_minus_restricted": greedy - restricted,
        "full_better": greedy < restricted,
        "restricted_better": restricted < greedy,
        "tie": np.isclose(greedy, restricted, atol=1e-8, rtol=0),
    })

df = pd.DataFrame(rows)

display(df[[
    "replicate",
    "ip_support_size",
    "greedy_full_loss",
    "greedy_restricted_loss",
    "ip_support_loss",
    "full_minus_restricted",
    "full_better",
    "restricted_better",
    "tie",
]])

print("Metadata:", data["metadata"])
print()
print("Mean full greedy loss:", df["greedy_full_loss"].mean())
print("Mean restricted greedy loss:", df["greedy_restricted_loss"].mean())
print("Mean IP support loss:", df["ip_support_loss"].mean())
print()
print("P(full greedy > restricted greedy):", (df["greedy_full_loss"] > df["greedy_restricted_loss"]).mean())
print("P(full greedy < restricted greedy):", (df["greedy_full_loss"] < df["greedy_restricted_loss"]).mean())
print("P(tie):", df["tie"].mean())
```

```python
import json
import numpy as np
import matplotlib.pyplot as plt

from isometry_pursuit.transformation import exponential_transformation

results_path = "results/ethanol_results.json"
replicate = 0

with open(results_path) as f:
    results = json.load(f)

data_file = results["metadata"]["data_file"]
X_tensor = np.load(data_file)

X = X_tensor[:, :, replicate]
X_t = exponential_transformation(X)

rep = results["replicates"][replicate]

ip_indices = rep["basis_pursuit_indices"]
greedy_full_indices = rep["greedy_gl_indices"]
greedy_restricted_indices = rep["greedy_gl_bp_indices"]

print("replicate:", replicate)
print("X shape:", X.shape)
print("transformed X shape:", X_t.shape)
print("IP support size:", len(ip_indices))
print("IP support:", ip_indices)
print("full greedy indices:", greedy_full_indices)
print("restricted greedy indices:", greedy_restricted_indices)

print()
print("full greedy transformed column norms:")
print(np.linalg.norm(X_t[:, greedy_full_indices], axis=0))

print()
print("restricted greedy transformed column norms:")
print(np.linalg.norm(X_t[:, greedy_restricted_indices], axis=0))

fig, ax = plt.subplots(figsize=(7, 7))

# all transformed columns
ax.scatter(
    X_t[0, :],
    X_t[1, :],
    s=12,
    alpha=0.25,
    label="all transformed columns",
)

# IP support
ax.scatter(
    X_t[0, ip_indices],
    X_t[1, ip_indices],
    s=70,
    facecolors="none",
    edgecolors="tab:blue",
    linewidths=1.8,
    label="IP support",
)

# full greedy
ax.scatter(
    X_t[0, greedy_full_indices],
    X_t[1, greedy_full_indices],
    s=140,
    marker="x",
    color="tab:red",
    linewidths=3,
    label="full greedy",
)

# restricted greedy
ax.scatter(
    X_t[0, greedy_restricted_indices],
    X_t[1, greedy_restricted_indices],
    s=120,
    marker="+",
    color="tab:green",
    linewidths=3,
    label="greedy restricted to IP",
)

for idx in sorted(set(ip_indices + greedy_full_indices + greedy_restricted_indices)):
    ax.annotate(
        str(idx),
        (X_t[0, idx], X_t[1, idx]),
        xytext=(4, 4),
        textcoords="offset points",
        fontsize=9,
    )

ax.axhline(0, color="black", linewidth=0.8, alpha=0.4)
ax.axvline(0, color="black", linewidth=0.8, alpha=0.4)
ax.set_aspect("equal", adjustable="box")
ax.set_xlabel("transformed coordinate 1")
ax.set_ylabel("transformed coordinate 2")
ax.legend()
ax.set_title(f"Ethanol replicate {replicate}: selected columns")
plt.show()
```

```python
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from isometry_pursuit.transformation import exponential_transformation

results_path = "results/ethanol_results.json"

with open(results_path) as f:
    results = json.load(f)

X_tensor = np.load(results["metadata"]["data_file"])

rows = []
for r in results["replicates"]:
    rows.append({
        "replicate": r["replicate"],
        "ip_support_size": r["support_cardinality"],
        "greedy_full_loss": r["greedy_gl_norm"],
        "greedy_restricted_loss": r["greedy_gl_bp_norm"],
        "full_minus_restricted": r["greedy_gl_norm"] - r["greedy_gl_bp_norm"],
        "ip_support": r["basis_pursuit_indices"],
        "greedy_full_indices": r["greedy_gl_indices"],
        "greedy_restricted_indices": r["greedy_gl_bp_indices"],
    })

df = pd.DataFrame(rows)

better = df[df["full_minus_restricted"] < 0].sort_values("full_minus_restricted")
display(better[[
    "replicate",
    "ip_support_size",
    "greedy_full_loss",
    "greedy_restricted_loss",
    "full_minus_restricted",
]])

replicate = int(better.iloc[0]["replicate"])
print("plotting strongest full-greedy win, replicate =", replicate)
```

```python
rep = results["replicates"][replicate]

X = X_tensor[:, :, replicate]
X_t = exponential_transformation(X)

ip_indices = rep["basis_pursuit_indices"]
greedy_full_indices = rep["greedy_gl_indices"]
greedy_restricted_indices = rep["greedy_gl_bp_indices"]

print("replicate:", replicate)
print("IP support:", ip_indices)
print("full greedy indices:", greedy_full_indices)
print("restricted greedy indices:", greedy_restricted_indices)
print("full greedy loss:", rep["greedy_gl_norm"])
print("restricted greedy loss:", rep["greedy_gl_bp_norm"])
print("full minus restricted:", rep["greedy_gl_norm"] - rep["greedy_gl_bp_norm"])

print()
print("full greedy transformed norms:", np.linalg.norm(X_t[:, greedy_full_indices], axis=0))
print("restricted greedy transformed norms:", np.linalg.norm(X_t[:, greedy_restricted_indices], axis=0))
print("IP support transformed norms:", np.linalg.norm(X_t[:, ip_indices], axis=0))

fig, ax = plt.subplots(figsize=(7, 7))

ax.scatter(
    X_t[0, :],
    X_t[1, :],
    s=12,
    alpha=0.22,
    color="gray",
    label="all transformed columns",
)

ax.scatter(
    X_t[0, ip_indices],
    X_t[1, ip_indices],
    s=90,
    facecolors="none",
    edgecolors="tab:blue",
    linewidths=2,
    label="IP support",
)

ax.scatter(
    X_t[0, greedy_full_indices],
    X_t[1, greedy_full_indices],
    s=170,
    marker="x",
    color="tab:red",
    linewidths=3,
    label="full greedy",
)

ax.scatter(
    X_t[0, greedy_restricted_indices],
    X_t[1, greedy_restricted_indices],
    s=150,
    marker="+",
    color="tab:green",
    linewidths=3,
    label="greedy restricted to IP",
)

important = sorted(set(ip_indices + greedy_full_indices + greedy_restricted_indices))
for idx in important:
    ax.annotate(
        str(idx),
        (X_t[0, idx], X_t[1, idx]),
        xytext=(5, 5),
        textcoords="offset points",
        fontsize=9,
    )

ax.axhline(0, color="black", linewidth=0.8, alpha=0.35)
ax.axvline(0, color="black", linewidth=0.8, alpha=0.35)
ax.set_aspect("equal", adjustable="box")
ax.set_xlabel("transformed coordinate 1")
ax.set_ylabel("transformed coordinate 2")
ax.set_title(
    f"Ethanol replicate {replicate}: "
    f"full greedy {rep['greedy_gl_norm']:.3g} vs restricted {rep['greedy_gl_bp_norm']:.3g}"
)
ax.legend()
plt.show()
```
