```python

def update_beta(X, y, z, nu):

    # Step 1: Unconstrained update
    beta_unproj = z - nu
    print('bj', beta_unproj)
    # Step 2: Projection onto the feasible set Xβ = y
    XXT_inv = np.linalg.inv(X @ X.T)  # (m x m) inverse
    projection = X.T @ XXT_inv @ (X @ beta_unproj - y)  # Projection step

    beta = beta_unproj - projection  # Final update
    print('beta', beta)
    return beta

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

%matplotlib inline

def admm_basis_pursuit_corrected(X, y, z, ρ=1.0, max_iter=100, tol=1e-6,    beta = None):
    
    
    n, p = X.shape
    if beta is None:
        beta = np.zeros(p)
    z_history, beta_history, u_history, v_history = [], [], [], []
    splitting_gap = beta - z
    constraint_violation = y - X @ beta
    
    for _ in range(max_iter):
        z_history.append(z.copy())
        beta_history.append(beta.copy())
        u_history.append(splitting_gap.copy())
        v_history.append(constraint_violation.copy())

        
        # β-update: Solve (XᵀX + ρI)β = Xᵀy + ρ(z - u) + Xᵀ(v)
        beta = update_beta(X, y, z, splitting_gap)

        # z-update: Soft-thresholding
        z_prev = z.copy()
        z = np.sign(beta + splitting_gap) * np.maximum(np.abs(beta + splitting_gap) - 1 / ρ, 0)

        # Dual updates
        splitting_gap += beta - z
        constraint_violation += y - X @ beta

    return beta, z, splitting_gap, beta_history, z_history, u_history, v_history

def plot_trajectories_3d(iteration, beta_history, z_history, u_history):
    fig = plt.figure(figsize=(8, 6))
    ax = fig.add_subplot(111, projection='3d')

    # Convert to NumPy arrays
    z_history = np.array(z_history)
    u_history = np.array(u_history)
    beta_history = np.array(beta_history)

    colors = {'z': 'blue', 'u': 'red', 'beta': 'purple'}

    # Function to plot dotted lines to hyperplanes
    def plot_dotted_lines(ax, point, color):
        x, y, z = point
        ax.plot([x, x], [y, y], [z, 0], linestyle='dotted', color=color, alpha=0.5)  # To XY plane
        ax.plot([x, x], [y, 0], [z, z], linestyle='dotted', color=color, alpha=0.5)  # To XZ plane
        ax.plot([x, 0], [y, y], [z, z], linestyle='dotted', color=color, alpha=0.5)  # To YZ plane

    # Plot final points with dotted lines
    for label, history, color in zip(['z', 'u', 'beta'], [z_history, u_history, beta_history], colors.values()):
        point = history[iteration]
        ax.scatter(*point, color=color, label=f'{label}', alpha=0.7, s = 100)
        plot_dotted_lines(ax, point, color)

    xx, yy = np.meshgrid(np.linspace(0,1, 10), np.linspace(0,1, 10))
    zz = (y - x[:, 0] * xx - x[:, 1] * yy) / x[:, 2]  # Solve for z in the plane equation
    ax.plot_surface(xx, yy, zz, alpha=0.1, color='gray')
    additional_point = z_history[iteration] - u_history[iteration] 
    
    ax.scatter(*additional_point, color='green', label=r'$z - u $', alpha=0.7, s = 100)
    plot_dotted_lines(ax, additional_point, 'green')
    
    # Set axis limits and ensure the origin is visible
    ax.set_xlim([-1, 1])
    ax.set_ylim([-1, 1])
    ax.set_zlim([-1, 1])

    # Draw X, Y, and Z axes
    ax.quiver(0, 0, 0, 1, 0, 0, color='black', arrow_length_ratio=0.1)  # X-axis
    ax.quiver(0, 0, 0, 0, 1, 0, color='black', arrow_length_ratio=0.1)  # Y-axis
    ax.quiver(0, 0, 0, 0, 0, 1, color='black', arrow_length_ratio=0.1)  # Z-axis

    ax.set_xlabel('X-axis')
    ax.set_ylabel('Y-axis')
    ax.set_zlabel('Z-axis')
    ax.set_title(f'Iteration {iteration}: Norm of Beta: {np.linalg.norm(beta_history[iteration]):.3f}, {beta_history[iteration]} ')
    
    ax.legend()
    ax.grid(False)  # Turn off the grid
    ax.view_init(elev=10, azim=45)

    plt.show()


# Parameters
p = 3
n = 1
x = np.asarray([[1., 1., 1.]])
rho = 1.0
max_iter = 10

# Run ADMM
y = np.ones(n)
# z = np.zeros(p)
z = np.asarray([1,0.,0.])
beta =  np.asarray([1,0,0])
# u = np.zeros(p)
splitting_gap = np.asarray([0.,0.,0.])
u = splitting_gap 
v = np.zeros(n)

beta_solution, z, u, beta_history, z_history, u_history, v_history = admm_basis_pursuit_corrected(x, y, z, beta= beta,ρ=rho, max_iter=max_iter)

# Create a plot for iterations 0 to 5
for iteration in range(max_iter):
    plot_trajectories_3d(iteration, beta_history, z_history, u_history)
```

```python
import numpy as np

# Set a seed for reproducibility
np.random.seed(0)

# Generate a 2 x 4 random Gaussian (multivariate normal) matrix
A = np.random.randn(2, 4)

# Compute Moore–Penrose pseudoinverse via SVD
A_pinv = np.linalg.pinv(A)

print("A:\n", A)
print("\nA^+ (pseudoinverse):\n", A_pinv)

# Verify properties:
# 1. A * A^+ ≈ I_2 (right inverse property, since A is full row rank with high prob)
print("\nA @ A^+ ≈ I_2:\n", A @ A_pinv)

# 2. A^+ * A is an orthogonal projector onto the row space of A
print("\nA^+ @ A (projector onto row space, 4x4):\n", A_pinv @ A)
```

```python

# Full SVD: A = U Σ V^T
U, s, Vt = np.linalg.svd(A, full_matrices=True)

# Invert nonzero singular values
# Build Σ^+ with shape 4 x 2 (transpose of Σ)
Sigma_pinv = np.zeros((Vt.shape[0], U.shape[0]))
for i in range(len(s)):
    if s[i] > 1e-12:   # tolerance to avoid blow-up
        Sigma_pinv[i, i] = 1.0 / s[i]

# Pseudoinverse: A^+ = V Σ^+ U^T
A_pinv_svd = Vt.T @ Sigma_pinv @ U.T

print("A:\n", A)
print("\nSingular values:", s)
print("\nA^+ from SVD:\n", A_pinv_svd)

# Check right inverse property (since A is wide and full row rank):
print("\nA @ A^+ ≈ I_2:\n", A @ A_pinv_svd)
```
