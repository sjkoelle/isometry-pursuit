```python
x,y,z = A.transpose()
x
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # noqa: F401

fig = plt.figure(figsize=(7, 7))
ax = fig.add_subplot(111, projection="3d")

# arrows from origin to (x,y,z)
ox = np.zeros_like(x)
oy = np.zeros_like(y)
oz = np.zeros_like(z)
ax.quiver(ox, oy, oz, x, y, z, length=1, normalize=False)

# points at tips
ax.scatter(x, y, z, s=20)

# label each arrow near the tip along its direction
for i, (xi, yi, zi) in enumerate(zip(x, y, z)):
    v = np.array([xi, yi, zi], dtype=float)
    n = np.linalg.norm(v)
    if n == 0:
        nx, ny, nz = 0.0, 0.0, 0.0
    else:
        nx, ny, nz = v / n
    off = 0.03
    ax.text(xi + off*nx, yi + off*ny, zi + off*nz, str(i),
            fontsize=14, ha="center", va="center")

# limits & aspect
m = 1.1 * np.max(np.abs(np.concatenate([x.ravel(), y.ravel(), z.ravel()])))
ax.set_xlim(-m, m)
ax.set_ylim(-m, m)
ax.set_zlim(-m, m)
try:
    ax.set_box_aspect((1, 1, 1))  # equal aspect (mpl >= 3.3)
except Exception:
    pass

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
ax.set_title("3D arrows with indices")
ax.view_init(elev=70, azim=110)  # e.g., look from above and from +y side
# optional: tweak the camera
# ax.view_init(elev=20, azim=35)

plt.show()
```
