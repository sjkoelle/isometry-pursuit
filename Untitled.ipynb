```python
import numpy as np
data = np.load('data/orthogonal/orthogonal_data_tensor.npy')
np.linalg.norm(data[0], axis = 0)
```

```python
A = data[0]
x,y = A
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 6))
ax.quiver(np.zeros_like(x), np.zeros_like(y), x, y,
          angles='xy', scale_units='xy', scale=1, width=0.005)
ax.scatter(x, y, s=20)

# Label each arrow with its index near the tip
for i, (xi, yi) in enumerate(zip(x, y)):
    # small offset along the vector direction to avoid overlap with the marker
    off = 0.03
    nx, ny = (xi, yi) if (xi == 0 and yi == 0) else (xi/np.hypot(xi, yi), yi/np.hypot(xi, yi))
    ax.text(xi + off*nx, yi + off*ny, str(i), fontsize=19,
            ha='center', va='center')

ax.set_aspect('equal', adjustable='box')
m = 1.1 * np.max(np.abs(A))
ax.set_xlim(-m, m)
ax.set_ylim(-m, m)
ax.grid(True, linestyle='--', alpha=0.4)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Arrows with indices')

plt.show()
```

```python
from isometry_pursuit.algorithm import isometry_pursuit

# Extract first replicate as 2D matrix
# New format: (n_replicates, D, n_samples) -> (D, n_samples)
X = data[0, :, :]

# Run isometry pursuit
constraint_violation = 1.99
result = isometry_pursuit(
    X,
    constraint_violation=constraint_violation,
)
```
