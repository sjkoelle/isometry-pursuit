```python
import numpy as np
import matplotlib.pyplot as plt

# epsilon values, avoiding exactly 0
eps = np.linspace(1e-4, 0.999, 1000)

s = np.sqrt(1 - eps**2)

sigma1 = np.sqrt(1 + s)
sigma2 = np.sqrt(1 - s)

dsigma1_dr = 0.5 * sigma1
dsigma2_dr = 0.5 * sigma2

plt.figure(figsize=(8, 5))

plt.plot(eps, sigma1, label=r"$\sigma_1(X_\epsilon(1))$")
plt.plot(eps, sigma2, label=r"$\sigma_2(X_\epsilon(1))$")
plt.plot(eps, dsigma1_dr, "--", label=r"$\partial \sigma_1/\partial r|_{r=1}$")
plt.plot(eps, dsigma2_dr, "--", label=r"$\partial \sigma_2/\partial r|_{r=1}$")

plt.xlabel(r"$\epsilon$")
plt.ylabel("value")
plt.title(r"Singular values and $r$-derivatives at $r=1$")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

eps = 0.2
r = np.linspace(0.2, 3.0, 1000)

Delta = np.sqrt((1 + r**2)**2 - 4 * eps**2 * r**2)

sigma1 = np.sqrt((1 + r**2 + Delta) / 2)
sigma2 = np.sqrt((1 + r**2 - Delta) / 2)

dsigma1_dr = (
    r / (2 * sigma1)
    * (1 + (1 + r**2 - 2 * eps**2) / Delta)
)

dsigma2_dr = (
    r / (2 * sigma2)
    * (1 - (1 + r**2 - 2 * eps**2) / Delta)
)

plt.figure(figsize=(8, 5))

plt.plot(r, dsigma1_dr, label=r"$\partial \sigma_1 / \partial r$")
plt.plot(r, dsigma2_dr, label=r"$\partial \sigma_2 / \partial r$")

plt.axvline(1, linestyle="--", linewidth=1, label=r"$r=1$")

plt.xlabel(r"$r$")
plt.ylabel("derivative")
plt.title(r"Derivatives of singular values with respect to $r$, $\epsilon=0.2$")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

eps = 0.2
r = np.linspace(0.2, 3.0, 1000)

Delta = np.sqrt((1 + r**2)**2 - 4 * eps**2 * r**2)

sigma1 = np.sqrt((1 + r**2 + Delta) / 2)
sigma2 = np.sqrt((1 + r**2 - Delta) / 2)

def g(t):
    return t + 1 / t

loss = g(sigma1) + g(sigma2)

plt.figure(figsize=(8, 5))
plt.plot(r, loss, label=r"$g(\sigma_1)+g(\sigma_2)$")
plt.axvline(1, linestyle="--", linewidth=1, label=r"$r=1$")

plt.xlabel(r"$r$")
plt.ylabel("loss")
plt.title(r"Loss vs $r$ for $\epsilon=0.2$, $g(t)=t+t^{-1}$")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

eps = 0.2
r = np.linspace(0.2, 3.0, 1000)

Delta = np.sqrt((1 + r**2)**2 - 4 * eps**2 * r**2)

sigma1 = np.sqrt((1 + r**2 + Delta) / 2)
sigma2 = np.sqrt((1 + r**2 - Delta) / 2)

def g(t):
    return 0.5 * t**2 - np.log(t)

loss = g(sigma1) + g(sigma2)

plt.figure(figsize=(8, 5))
plt.plot(r, loss, label=r"$g(\sigma_1)+g(\sigma_2)$")
plt.axvline(1, linestyle="--", linewidth=1, label=r"$r=1$")

plt.xlabel(r"$r$")
plt.ylabel("loss")
plt.title(r"Loss vs $r$ for $\epsilon=0.2$, $g(t)=t^2/2-\log t$")
plt.legend()
plt.grid(True)
plt.show()
```
